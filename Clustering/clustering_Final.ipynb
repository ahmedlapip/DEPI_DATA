```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

# قراءة البيانات
try:
    data = pd.read_csv('averaged_attributes_by_season.csv')
except FileNotFoundError:
    raise FileNotFoundError("ملف 'averaged_attributes_by_season.csv' غير موجود! شغلي الكود السابق لإنشاء الملف.")

# اختيار السمات للـ Clustering
features = ['PRECTOTCORR_avg_autumn', 'PRECTOTCORR_avg_spring', 'PRECTOTCORR_avg_summer', 'PRECTOTCORR_avg_winter',
            'RH2M_avg_autumn', 'RH2M_avg_spring', 'RH2M_avg_summer', 'RH2M_avg_winter',
            'T2M_avg_autumn', 'T2M_avg_spring', 'T2M_avg_summer', 'T2M_avg_winter']

# تحضير البيانات
X = data[features]
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# تطبيق K-Means بـ 4 Clusters
kmeans = KMeans(n_clusters=4, random_state=42, n_init=15)
clusters = kmeans.fit_predict(X_scaled)

# إضافة الـ Clusters للبيانات
data['Cluster'] = clusters

# حفظ البيانات مع التصنيفات في نفس الفايل
data.to_csv('averaged_attributes_by_season.csv', index=False)
print("\nتم حفظ البيانات مع التصنيفات في ملف 'averaged_attributes_by_season.csv'!")
```

```python
import pandas as pd

# قراءة البيانات
try:
    data = pd.read_csv('Plants.csv')
except FileNotFoundError:
    raise FileNotFoundError("ملف 'plant.csv' غير موجود! تأكدي من توفر الملف.")

# تحديد الأعمدة المطلوبة
attributes = ['WS2M', 'RH2M', 'T2M', 'TSOIL1', 'TSOIL4', 'PRECTOTCORR', 'GWETROOT', 'GWETTOP', 'ALLSKY_SFC_SW_DWN']

# التحقق من وجود الأعمدة المطلوبة
if not all(col in data.columns for col in attributes):
    raise ValueError("البيانات يجب أن تحتوي على الأعمدة: " + ", ".join(attributes))

# دالة لتقسيم القيم إلى min و max
def split_min_max(value):
    if pd.isna(value) or not isinstance(value, str):
        return pd.Series([None, None])
    try:
        min_val, max_val = value.split('-')
        return pd.Series([float(min_val), float(max_val)])
    except (ValueError, AttributeError):
        return pd.Series([None, None])

# تطبيق التقسيم على كل عمود
for attr in attributes:
    min_max = data[attr].apply(split_min_max)
    data[attr + '_min'] = min_max[0]
    data[attr + '_max'] = min_max[1]

# حذف الأعمدة الأصلية
data = data.drop(columns=attributes)

# حفظ النتيجة في فايل جديد
data.to_csv('plant_split_min_max.csv', index=False)
print("\nتم تقسيم القيم إلى min و max وحفظ النتيجة في ملف 'plant_split_min_max.csv' بدون الأعمدة الأصلية!")
```

```python
import pandas as pd

# قراءة البيانات
try:
    plants_data = pd.read_csv('plant_split_min_max.csv')
except FileNotFoundError:
    raise FileNotFoundError("ملف 'plant_split_min_max.csv' غير موجود! تأكدي من توفر الملف.")

try:
    clusters_data = pd.read_csv('cluster_attributes_min_max_by_season.csv')
except FileNotFoundError:
    raise FileNotFoundError("ملف 'cluster_attributes_min_max_by_season.csv' غير موجود! تأكدي من توفر الملف.")

# تحديد السمات المطلوبة
attributes = ['WS2M', 'RH2M', 'T2M', 'TSOIL1', 'TSOIL4', 'PRECTOTCORR', 'GWETROOT', 'GWETTOP', 'ALLSKY_SFC_SW_DWN']
plant_cols = [f"{attr}_min" for attr in attributes] + [f"{attr}_max" for attr in attributes]
cluster_cols = ['Cluster', 'Season'] + [f"{attr}_percentile_5" for attr in attributes] + [f"{attr}_percentile_95" for attr in attributes]

# التحقق من وجود الأعمدة المطلوبة
if not all(col in plants_data.columns for col in plant_cols + ['Plant']):
    raise ValueError("الفايل 'plant_split_min_max.csv' يجب أن يحتوي على الأعمدة: " + ", ".join(plant_cols + ['Plant']))

if not all(col in clusters_data.columns for col in cluster_cols):
    raise ValueError("الفايل 'cluster_attributes_min_max_by_season.csv' يجب أن يحتوي على الأعمدة: " + ", ".join(cluster_cols))

# دالة للتحقق من التوافق بين نطاق النبات ونطاق الـ Cluster
def check_compatibility(plant_row, cluster_row, attributes):
    for attr in attributes:
        plant_min = plant_row[f"{attr}_min"]
        plant_max = plant_row[f"{attr}_max"]
        cluster_min = cluster_row[f"{attr}_percentile_5"]
        cluster_max = cluster_row[f"{attr}_percentile_95"]
        
        # التحقق من أن القيم ليست NaN
        if pd.isna(plant_min) or pd.isna(plant_max) or pd.isna(cluster_min) or pd.isna(cluster_max):
            return False
        
        # التحقق من التوافق: النطاقات بتتقاطع لو الـ max بتاع واحد أكبر من الـ min بتاع التاني
        if plant_max < cluster_min or plant_min > cluster_max:
            return False
    
    return True

# قائمة لتخزين النتايج
results = []

# مقارنة كل نبات مع كل Cluster وكل موسم
for plant_idx, plant_row in plants_data.iterrows():
    plant_name = plant_row['Plant']
    compatible_clusters = []
    
    for cluster_idx, cluster_row in clusters_data.iterrows():
        cluster = cluster_row['Cluster']
        season = cluster_row['Season']
        
        if check_compatibility(plant_row, cluster_row, attributes):
            compatible_clusters.append(f"Cluster {cluster} in {season}")
    
    # إضافة النتيجة
    if compatible_clusters:
        results.append({'Plant': plant_name, 'Compatible_Clusters': ', '.join(compatible_clusters)})
    else:
        results.append({'Plant': plant_name, 'Compatible_Clusters': 'None'})

# إنشاء DataFrame للنتايج
results_df = pd.DataFrame(results)

# حفظ النتيجة في فايل جديد
results_df.to_csv('plant_cluster_compatibility.csv', index=False)
print("\nتم مقارنة النباتات مع الـ Clusters وحفظ النتيجة في ملف 'plant_cluster_compatibility.csv'!")
```

```python
import pandas as pd

# قراءة البيانات
try:
    mapping_data = pd.read_csv('cluster_season_plant_mapping.csv')
except FileNotFoundError:
    raise FileNotFoundError("ملف 'cluster_season_plant_mapping.csv' غير موجود! تأكدي من توفر الملف.")

try:
    plants_data = pd.read_csv('plant_split_min_max.csv')
except FileNotFoundError:
    raise FileNotFoundError("ملف 'plant_split_min_max.csv' غير موجود! تأكدي من توفر الملف.")

try:
    clusters_data = pd.read_csv('cluster_attributes_min_max_by_season.csv')
except FileNotFoundError:
    raise FileNotFoundError("ملف 'cluster_attributes_min_max_by_season.csv' غير موجود! تأكدي من توفر الملف.")

# تحديد السمات المطلوبة
attributes = ['WS2M', 'RH2M', 'T2M', 'TSOIL1', 'TSOIL4', 'PRECTOTCORR', 'GWETROOT', 'GWETTOP', 'ALLSKY_SFC_SW_DWN']
plant_cols = [f"{attr}_min" for attr in attributes] + [f"{attr}_max" for attr in attributes]
cluster_cols = ['Cluster', 'Season'] + [f"{attr}_percentile_5" for attr in attributes] + [f"{attr}_percentile_95" for attr in attributes]

# التحقق من وجود الأعمدة المطلوبة
if not all(col in mapping_data.columns for col in ['Cluster', 'Season', 'Plant']):
    raise ValueError("الفايل 'cluster_season_plant_mapping.csv' يجب أن يحتوي على الأعمدة: Cluster, Season, Plant")

if not all(col in plants_data.columns for col in plant_cols + ['Plant']):
    raise ValueError("الفايل 'plant_split_min_max.csv' يجب أن يحتوي على الأعمدة: " + ", ".join(plant_cols + ['Plant']))

if not all(col in clusters_data.columns for col in cluster_cols):
    raise ValueError("الفايل 'cluster_attributes_min_max_by_season.csv' يجب أن يحتوي على الأعمدة: " + ", ".join(cluster_cols))

# دالة للتحقق من التوافق وتحديد >, <, أو =
def check_attribute_level(plant_row, cluster_row, attr):
    plant_min = plant_row[f"{attr}_min"]
    plant_max = plant_row[f"{attr}_max"]
    cluster_min = cluster_row[f"{attr}_percentile_5"]
    cluster_max = cluster_row[f"{attr}_percentile_95"]
    
    if pd.isna(plant_min) or pd.isna(plant_max) or pd.isna(cluster_min) or pd.isna(cluster_max):
        return 'N/A'
    
    # تحديد الحالة بناءً على المقارنة
    if plant_max < cluster_min:  # النبات يحتاج أقل من المتوفر
        return '< Deficit'
    elif plant_min > cluster_max:  # النبات يحتاج أعلى مما هو متوفر
        return '> Excess'
    else:  # فيه تقاطع أو تطابق
        return '= Suitable'

# إنشاء DataFrame جديد بنفس البيانات من mapping_data
results_df = mapping_data.copy()

# إضافة أعمدة السمات مع قيم افتراضية
for attr in attributes:
    results_df[attr] = None

# معالجة كل صف في mapping_data
for idx, row in results_df.iterrows():
    cluster = row['Cluster']
    season = row['Season']
    plant_name = row['Plant']
    
    # استرجاع بيانات النبات
    plant_row = plants_data[plants_data['Plant'] == plant_name].iloc[0]
    
    # استرجاع بيانات الـ Cluster و Season
    cluster_row = clusters_data[(clusters_data['Cluster'] == cluster) & (clusters_data['Season'] == season)].iloc[0]
    
    # التحقق من مستوى كل سمة
    for attr in attributes:
        level = check_attribute_level(plant_row, cluster_row, attr)
        results_df.at[idx, attr] = level

# ترتيب الأعمدة
results_df = results_df[['Cluster', 'Season', 'Plant'] + attributes]

# حفظ النتيجة في فايل CSV
results_df.to_csv('cluster_plant_attributes_compatibility.csv', index=False)
print("\nتم إنشاء جدول التوافق مع علامات > Excess, < Deficit, = Suitable لكل سمة وحفظ النتيجة في ملف 'cluster_plant_attributes_compatibility.csv'!")
```
