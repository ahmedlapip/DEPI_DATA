```python
import folium
import numpy as np
import pandas as pd
from folium.plugins import HeatMap
from folium import plugins
import ipywidgets as widgets
from IPython.display import display, clear_output
import matplotlib.colors as mcolors
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
import calendar
from IPython.display import display, Markdown
from pathlib import Path
import plotly.colors
import random
```

```python
df_Large=pd.read_csv("data/fullWithLocations_FINAL_large.csv",index_col=False)

```

```python
display(df_Large.columns.sort_values())
display(df_Small.columns.sort_values())
```

```python
Group_large=['EVLAND', 'GWETPROF',
       'GWETROOT', 'GWETTOP', 'PRECSNO', 'PRECTOTCORR', 'QV2M', 'RH2M', 'RHOA','T10M', 'T10M_MAX', 'T10M_MIN', 'T2M', 'T2M_MAX', 'T2M_MIN', 'TO3',
       'TS', 'TSOIL1', 'TSOIL2', 'TSOIL3', 'TSOIL4', 'TSOIL5', 'TSOIL6','TS_MAX', 'TS_MIN', 'WD2M', 'WD50M', 'WS2M', 'WS2M_MAX', 'WS2M_MIN',
       'WS50M', 'WS50M_MAX', 'WS50M_MIN', 'Z0M']

Group_small=['AIRMASS', 'ALLSKY_SFC_SW_DWN', 'ALLSKY_SFC_UVA', 'ALLSKY_SFC_UVB',
       'ALLSKY_SFC_UV_INDEX', 'CLOUD_AMT', 'CLOUD_AMT_DAY', 'CLOUD_AMT_NIGHT',
       'CLRSKY_DAYS', 'City', 'DOY', 'Date', 
       'MIDDAY_INSOL', 'PSH', 'PW']

df_Mapped=pd.read_csv("data/dataMapping.csv")
```

```python
display(df_Large[Group_large].describe())
display(df_Large[Group_large].info())
```

```python
display(df_Small[Group_small].describe())
display(df_Small[Group_small].info())
```

```python
df_Large['Month_Name'] = df_Large['Month'].apply(lambda x: calendar.month_abbr[x]) 
df_Small['Month_Name'] = df_Small['Month'].apply(lambda x: calendar.month_abbr[x]) 
```

```python
attribute_options = df_Mapped["Code"]
attribute_selector = widgets.Dropdown(
    options=attribute_options,
    value="PRECTOTCORR",
    description="Attribute:"
)
display(attribute_selector)
```

```python
display(attribute_selector.value)
```

```python
attribute=attribute_selector.value
Att_Name = df_Mapped.loc[df_Mapped["Code"] == attribute, "Name"].iloc[0]
color_palette = plotly.colors.qualitative.Safe  
color_UN = random.choice(color_palette)
base_dir = Path(f"resources/{attribute}")
base_dir.mkdir(parents=True, exist_ok=True)
```

```python

Total_info = pd.DataFrame(columns=["Attribute", "NULLs_Count", "Zeros_Percentage"])
for Attribute in Group_large + Group_small:
    if Attribute in df_Large:
        new_row = pd.DataFrame([{
            "Attribute": Attribute,
            "NULLs_Count": df_Large[Attribute].isnull().sum(),
            "Zeros_Percentage": (df_Large[Attribute] == 0).mean() * 100  
        }])
        Total_info = pd.concat([Total_info, new_row], ignore_index=True)

    elif Attribute in df_Small:
        new_row = pd.DataFrame([{
            "Attribute": Attribute,
            "NULLs_Count": df_Small[Attribute].isnull().sum(),
            "Zeros_Percentage": (df_Small[Attribute] == 0).mean() * 100  
        }])
        Total_info = pd.concat([Total_info, new_row], ignore_index=True)

display(Total_info.sort_values(by="Zeros_Percentage"))
```

```python
if attribute in Group_large:
    df=df_Large.copy()
else :
    df =df_Small.copy()    
```

```python
display(Markdown(f"<h1 style='color:{color_UN}; font-weight:bold; font-size: 40px;'>Histogram FOR {Att_Name} PER MONTH</h1>"))
```

```python
for YEAR in df["YEAR"].unique():
    dff = df[df["YEAR"] == YEAR]  

    fig = px.histogram(dff, 
                       x=attribute, 
                       nbins=50,#50  
                       facet_col="Month_Name", 
                       facet_col_wrap=5,  # Wrap into rows of 3 columns
                       title=f"{Att_Name} for YEAR={YEAR}",  
                       labels={f"{Att_Name}": f"{Att_Name}", "count": "Frequency"},
                       color_discrete_sequence=["royalblue"],  
                       opacity=0.75,  
                      )

    fig.update_layout(
        template="plotly_white",
        xaxis_title=Att_Name,
        yaxis_title="Frequency",
        hovermode="x unified"
    )

    fig.show()
    fig.write_html(f"resources/{attribute}/histogram_{Att_Name}_byMonth_Year{YEAR}.html")
```

```python
display(Markdown(f"<h1 style='color:{color_UN}; font-weight:bold; font-size: 40px;'>Pie Chart FOR {Att_Name} PER MONTH</h1>"))
```

```python
df_avg = df.groupby(['YEAR', 'Month_Name'], as_index=False)[attribute].mean()

for year in df_avg["YEAR"].unique():
    year_data = df_avg[df_avg['YEAR'] == year]
    
    fig = px.pie(year_data, 
                names='Month_Name',  
                values=attribute,  
                title=f"Average {Att_Name} per Month in {year}",
                hole=0.3)  
    fig.update_traces(textinfo='label+percent')

    fig.show()
```

```python
display(Markdown(f"<h1 style='color:{color_UN}; font-weight:bold; font-size: 40px;'>BAR GRAPH FOR {Att_Name} PER MONTH</h1>"))
```

```python
for YEAR in df["YEAR"].unique():
    dff = df[df["YEAR"] == YEAR]
    all_months = pd.DataFrame({"Month": range(1, 13)}) 
    month_stats = dff.groupby("Month")[attribute].agg(["min", "max", "mean"]).reset_index()
    month_stats = all_months.merge(month_stats, on="Month", how="left").fillna(0)  

    month_stats_melted = month_stats.melt(id_vars="Month", var_name="Statistic", value_name=attribute)

    fig = px.bar(month_stats_melted, 
                x="Month", 
                y=attribute, 
                text=attribute,  
                title=f"Min, Max, and Mean {Att_Name} per Month for {YEAR}",
                labels={f"{attribute}": f"{Att_Name}", "Month": "Month"},
                color="Statistic",
                barmode="group",   
                color_discrete_map={"min": "blue", "max": "red", "mean": "green"}  
                )

    for trace in fig.data:
        if isinstance(trace, go.Bar):
            trace.texttemplate = "%{y:.2f}"
            trace.textposition = "outside"

    for stat, color in zip(["min", "max", "mean"], ["blue", "red", "green"]):
        fig.add_trace(go.Scatter(
            x=month_stats["Month"], 
            y=month_stats[stat], 
            mode="lines+markers", 
            name=f"{stat} Trend",
            line=dict(color=color, width=2, dash="dash"),
            marker=dict(size=6)
        ))

    fig.update_layout(
        xaxis=dict(tickmode="linear", dtick=1, tickangle=0), 
        template="plotly_white",
        hovermode="x",
        yaxis_title=attribute
    )
    
    fig.show()

    fig.write_html(f"resources/{attribute}/months_MinMaxMean_Barplot_Year{YEAR}.html")
```

```python
display(Markdown(f"<h1 style='color:{color_UN}; font-weight:bold; font-size: 40px;'>BAR GRAPH FOR {Att_Name} PER CITY</h1>"))
```

```python
for YEAR in df["YEAR"].unique():
    
    city_stats = df[df["YEAR"] == YEAR].groupby("City")[attribute].agg(["min", "max", "mean"]).reset_index()

    city_stats_melted = city_stats.melt(id_vars="City", var_name="Statistic", value_name=attribute)

    fig = px.bar(city_stats_melted, 
                 x="City", 
                 y=attribute, 
                 color="Statistic", 
                 barmode="group",
                 title=f"Min, Max, and Mean {Att_Name} per City for {YEAR}",
                 labels={f"{attribute}": f"{Att_Name}", "City": "City"},
                 color_discrete_map={"min": "#1f77b4", "max": "#ff7f0e", "mean": "#2ca02c"}  
                )

    for trace in fig.data:
        if isinstance(trace, go.Bar):
            trace.texttemplate = "%{y:.2f}"
            trace.textposition = "outside"

    for stat, color in zip(["min", "max", "mean"], ["#1f77b4", "#ff7f0e", "#2ca02c"]):
        fig.add_trace(go.Scatter(
            x=city_stats["City"], 
            y=city_stats[stat], 
            mode="lines+markers", 
            name=f"{stat} Trend",
            line=dict(color=color, width=2, dash="dash"),
            marker=dict(size=6)
        ))
    fig.update_layout(
        legend_title_text="Statistic",
        xaxis_title="City",
        yaxis_title=Att_Name,
        xaxis=dict(tickangle=270),
        template="plotly_white",
        hovermode="x"
    )

    fig.show()
    fig.write_html(f"resources/{attribute}/Cities_MinMaxMean_Barplot_Year{YEAR}.html")
```
